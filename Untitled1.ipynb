```python
#aggregate function = redueces a set of values into a single summary value used to summarise and analyze data often used with the groupby()
marketing=pd.read_csv("https://raw.githubusercontent.com/AmeyaPatil1/marketing_campaign_messy_data/refs/heads/main/marketing_campaign_data_messy.csv")
#for whole data frame
print(marketing.mean(numeric_only=True))#mean
print(marketing.sum(numeric_only=True))#sum
print(marketing.min(numeric_only=True))#min
print(marketing.max(numeric_only=True))#max
print(marketing.count())#count

#for a single program
print(marketing["Impressions"].mean(numeric_only=True))#mean
print(marketing['Clicks '].sum(numeric_only=True))#sum
print(marketing["Clicks"].min(numeric_only=True))#min
print(marketing["Conversions"].max(numeric_only=True))#max
print(marketing.count())#count

# group by
Group=marketing.groupby("Channel")
print(Group["Channel"].sum())
```

```python
# data cleaning:-for cleaning messy data and and removing irrelavent things
# drop
drop=marketing.drop(columns=["Channel","Clicks"])
print(drop)

#handling missing data
s=marketing.fillna({"Active":"none"})
print(s)
# fixing
d=marketing["Active"].replace({"0":"pp"})
print(d)
```
